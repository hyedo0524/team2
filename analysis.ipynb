```python
def load_dataset(filename):
    BASE_DIR = './data'
    EXTENSION = '.csv'
    TARGET_PATH = f'{BASE_DIR}/{filename}{EXTENSION}'
    try:
        return pd.read_csv(TARGET_PATH)
    except FileNotFoundError:
        return None
    except UnicodeDecodeError:
        return pd.read_csv(TARGET_PATH, encoding='euc-kr')
```

```python
화물작업종류 = ['양하', '적하', '양적하']

화물작업중앙값 = df_inner.loc[
    df_inner['입항목적'].isin(화물작업종류),
    '예정대기시간'
].median()

화물작업데이터 = df_inner[
    df_inner['입항목적'].isin(화물작업종류)
]

화물작업중앙값초과건수 = (
    화물작업데이터['예정대기시간'] > 화물작업중앙값
).sum()

# 화물작업 중에서 어느정도나 중앙값보다 큰지 확인한다

round((화물작업중앙값초과건수 / 화물작업데이터.shape[0]) * 100 , 2)
```

```python
선박유지보수종류 = ['선박수리', '급유', '선용품적재']

선박유지보수중앙값 = df_inner.loc[
    df_inner['입항목적'].isin(선박유지보수종류),
    '예정대기시간'
].median()

선박유지보수데이터 = df_inner[
    df_inner['입항목적'].isin(선박유지보수종류)
]

선박유지보수중앙값초과건수 = (
    화물작업데이터['예정대기시간'] > 선박유지보수중앙값
).sum()

# 화물작업 중에서 어느정도나 중앙값보다 큰지 확인한다

round((선박유지보수중앙값초과건수 / 화물작업데이터.shape[0]) * 100 , 2)
```

```python
인력및운항종류 = ['승무원교대', '단순경유']

인력및운항종류중앙값 = df_inner.loc[
    df_inner['입항목적'].isin(인력및운항종류),
    '예정대기시간'
].median()

인력및운항데이터 = df_inner[
    df_inner['입항목적'].isin(인력및운항종류)
]

인력및운항중앙값초과건수 = (
    화물작업데이터['예정대기시간'] > 인력및운항종류중앙값
).sum()

# 화물작업 중에서 어느정도나 중앙값보다 큰지 확인한다

round((인력및운항중앙값초과건수 / 화물작업데이터.shape[0]) * 100 , 2)
```
